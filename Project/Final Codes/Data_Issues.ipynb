```python
import pandas as pd
import numpy as np
```

## Duplicate Product IDs

```python
file_loc_dict = {}

with open('env.txt','r') as f:
  f_reader = f.readlines()

for x in f_reader:
  key = x.split("=")[0]
  val = x.split("=")[1].split('\n')[0]
  file_loc_dict[key] = val

  
```

```python
# loc = "D:\github_windows\DE_CS_202309\Project"
# file = loc+"\\"+"Final.xlsx"

loc = file_loc_dict['data_iss_loc']
file = file_loc_dict['data_iss_file']
```

```python
df = pd.read_excel(file)
df.head()
```

```python
df_prod = df[['Product ID','Category','Sub-Category','Product Name']]
df_prod.info()
```

```python
df_prod = df_prod.drop_duplicates()
df_prod.info()
```

```python
df_prod.head()
```

```python
np.count_nonzero(df_prod['Product Name'].unique())
```

```python
df_prod[df_prod["Product ID"].duplicated()]
```
